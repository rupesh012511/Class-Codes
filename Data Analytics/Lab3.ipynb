```python
import pandas as pd

print("=== Missing Data Handling Program ===")

file_path = "students_marks_raw.csv"

try:
    df = pd.read_csv(file_path)
    print("\nFile loaded successfully!\n")
except Exception as e:
    print("Error: Could not read the file.")
    print(e)
    exit()

print("Missing values in each column:")
print(df.isnull().sum())


# 1. Mean Imputation
mean_df = df.copy()
for col in mean_df.select_dtypes(include="number").columns:
    mean_df[col] = mean_df[col].fillna(mean_df[col].mean())
mean_df.to_csv("mean_imputed.csv", index=False)


# 2. Median Imputation
median_df = df.copy()
for col in median_df.select_dtypes(include="number").columns:
    median_df[col] = median_df[col].fillna(median_df[col].median())
median_df.to_csv("median_imputed.csv", index=False)


# 3. Mode Imputation
mode_df = df.copy()
for col in mode_df.columns:
    mode_df[col] = mode_df[col].fillna(mode_df[col].mode()[0])
mode_df.to_csv("mode_imputed.csv", index=False)


# 4. Forward Fill
ffill_df = df.ffill()
ffill_df.to_csv("forward_filled.csv", index=False)

# 5. Backward Fill
bfill_df = df.bfill()
bfill_df.to_csv("backward_filled.csv", index=False)


print("\nAll methods applied successfully!")
print("Files created:")
print("mean_imputed.csv")
print("median_imputed.csv")
print("mode_imputed.csv")
print("forward_filled.csv")
print("backward_filled.csv")
```
